### tf-idf 
tf : term frequency

idf : inverse document frequence

TF = termfreq in doc / total term in doc

IDF = loge(total no. of doc in corpus) / termfreq in corpus

TF-IDF = TF * IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
  "I love  pizza pizza pizza",
  "Pizza is the best Pizza",
  "I love pasta",
  "Pasta is great"
]

# model
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

print("Vocabulary:", vectorizer.get_feature_names_out()) # exclude stop words
print("\n BoW Matrix:\n\n", X.toarray()) #count frequency
X.shape
```
